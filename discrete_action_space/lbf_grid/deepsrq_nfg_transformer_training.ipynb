# Deep SRQ NfgTransformer Training

Training-only notebook for Deep SRQ with the `nfg_transformer_sre` amortized neural stage-game solver. Each epsilon cell runs all configured LBF scenarios and saves full-resume checkpoints plus reward plots.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "lbf_grid":
    REPO_ROOT = ROOT.parents[1]
elif (ROOT / "discrete_action_space" / "lbf_grid").exists():
    REPO_ROOT = ROOT
else:
    REPO_ROOT = ROOT.parents[1]

for path in [REPO_ROOT, REPO_ROOT / "discrete_action_space", REPO_ROOT / "discrete_action_space" / "lbf_grid"]:
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.lbf_grid.robust_notebook_utils import (
    ROBUST_EPSILONS,
    robust_lbf_scenarios,
    train_deepsrq_nfg_transformer_for_epsilon,
)
from discrete_action_space.lbf_grid.notebook_eval import display_training_reward_plots

# SCENARIOS = robust_lbf_scenarios()
# for scenario in SCENARIOS:
#     print(f"{scenario.key}: {scenario.name}")
SCENARIOS = [robust_lbf_scenarios()[-1]]

def newly_trained_records(records):
    return [record for record in records if record.get("status") != "skipped_existing"]
```

```python
N_EPISODES = 5000
USE_GPU = True
BASE_SEED = 2025
EVAL_INTERVAL = 100
EVAL_EPISODES_DURING_TRAINING = 10
NUM_ENVS = 32
EVAL_NUM_ENVS = 8
SKIP_EXISTING_TRAINING = True
NFG_TRANSFORMER_CHECKPOINT_PATH = (
    REPO_ROOT
    / "discrete_action_space"
    / "sre_solvers"
    / "nfg_transformer"
    / "nfg_sre_checkpoints"
    / "nfg_sre_lbf3_online.pt"
)
NFG_TRANSFORMER_HYPERPARAMETER_OVERRIDES = {
    "agent": {
        "train_every": 4,
        "target_equilibrium_update_steps": 4,
        "sre_policy_cache_enabled": False,
    },
    "nfg_transformer": {
        "checkpoint_path": NFG_TRANSFORMER_CHECKPOINT_PATH,
        "fallback_enabled": False,
        "accept_exploitability_tol": None,
        "compute_exploitability_diagnostics": True,
        "device": "cuda" if USE_GPU else "cpu",
    },
}

print(f"NfgTransformer checkpoint: {NFG_TRANSFORMER_CHECKPOINT_PATH}")
print(f"Checkpoint exists: {NFG_TRANSFORMER_CHECKPOINT_PATH.exists()}")
```

## Robust epsilon 0.01


```python
deep_srq_nfg_transformer_eps_0_01_stats = train_deepsrq_nfg_transformer_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    scenarios=SCENARIOS,
    base_seed=BASE_SEED + 1,
    use_gpu=USE_GPU,
    eval_interval=EVAL_INTERVAL,
    eval_episodes=EVAL_EPISODES_DURING_TRAINING,
    num_envs=NUM_ENVS,
    eval_num_envs=EVAL_NUM_ENVS,
    hyperparameter_overrides=NFG_TRANSFORMER_HYPERPARAMETER_OVERRIDES,
    repo_root=REPO_ROOT,
    skip_existing=SKIP_EXISTING_TRAINING,
)
deep_srq_nfg_transformer_eps_0_01_training_figs = display_training_reward_plots(newly_trained_records(deep_srq_nfg_transformer_eps_0_01_stats.values()))
```
