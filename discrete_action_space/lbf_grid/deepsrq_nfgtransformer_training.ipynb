# Deep SRQ NfgTransformer Training

Training-only notebook for Deep SRQ with the `nfg_transformer_sre` stage-game solver. Each epsilon cell runs all configured LBF scenarios and saves full-resume checkpoints plus reward plots.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "lbf_grid":
    REPO_ROOT = ROOT.parents[1]
elif (ROOT / "discrete_action_space" / "lbf_grid").exists():
    REPO_ROOT = ROOT
else:
    REPO_ROOT = ROOT.parents[1]

for path in [REPO_ROOT, REPO_ROOT / "discrete_action_space", REPO_ROOT / "discrete_action_space" / "lbf_grid"]:
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.lbf_grid.robust_notebook_utils import (
    DEFAULT_EVAL_EPISODES,
    DEFAULT_NFG_TRANSFORMER_CHECKPOINT,
    ROBUST_EPSILONS,
    evaluate_deepsrq_suite_for_epsilon,
    robust_lbf_scenarios,
    train_deepsrq_nfgtransformer_for_epsilon,
)

SCENARIOS = robust_lbf_scenarios()
for scenario in SCENARIOS:
    print(f"{scenario.key}: {scenario.name}")
NUM_ENVS = 32
EVAL_NUM_ENVS = NUM_ENVS

```

```python
import torch

N_EPISODES = 3000
USE_GPU = True
BASE_SEED = 2025
EVAL_INTERVAL = 100
EVAL_EPISODES_DURING_TRAINING = 10
NFG_TRANSFORMER_CHECKPOINT = DEFAULT_NFG_TRANSFORMER_CHECKPOINT
NFG_ACCEPT_GAP = 0.1

print(f"NfgTransformer checkpoint: {NFG_TRANSFORMER_CHECKPOINT}")
print(f"Checkpoint exists: {Path(NFG_TRANSFORMER_CHECKPOINT).exists()}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01

```python
deep_srq_eps_0_01_stats = train_deepsrq_nfgtransformer_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    scenarios=SCENARIOS,
    nfg_checkpoint_path=NFG_TRANSFORMER_CHECKPOINT,
    nfg_accept_gap=NFG_ACCEPT_GAP,
    base_seed=BASE_SEED + 1,
    use_gpu=USE_GPU,
    eval_interval=EVAL_INTERVAL,
    eval_episodes=EVAL_EPISODES_DURING_TRAINING,
    repo_root=REPO_ROOT,
    num_envs=NUM_ENVS,
    eval_num_envs=EVAL_NUM_ENVS,
)
deep_srq_eps_0_01_stats
```
