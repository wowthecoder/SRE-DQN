# Deep SRQ Smoothed SRED Gradient Training

Training-only notebook for Deep SRQ with the `sred_gd_sre` stage-game solver. Each epsilon cell runs all configured LBF scenarios and saves full-resume checkpoints plus reward plots.

```python
from pathlib import Path
import sys
import torch

ROOT = Path.cwd()
if ROOT.name == "lbf_grid":
    REPO_ROOT = ROOT.parents[1]
elif (ROOT / "discrete_action_space" / "lbf_grid").exists():
    REPO_ROOT = ROOT
else:
    REPO_ROOT = ROOT.parents[1]

for path in [REPO_ROOT, REPO_ROOT / "discrete_action_space", REPO_ROOT / "discrete_action_space" / "lbf_grid"]:
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.lbf_grid.robust_notebook_utils import (
    ROBUST_EPSILONS,
    robust_lbf_scenarios,
    train_deepsrq_sred_gradient_for_epsilon,
)
from discrete_action_space.lbf_grid.notebook_eval import display_training_reward_plots

SCENARIOS = robust_lbf_scenarios()
for scenario in SCENARIOS:
    print(f"{scenario.key}: {scenario.name}")
```

```python
N_EPISODES = 5000
USE_GPU = True
BASE_SEED = 2025
EVAL_INTERVAL = 100
EVAL_EPISODES_DURING_TRAINING = 10
SRED_DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Keep the notebook responsive. Increase `sred_max_iters` and `sre_num_repeats`
# for slower final-quality sweeps after the training path is behaving.
SRED_HYPERPARAMETER_OVERRIDES = {
    "batch_size": 64,
    "train_every": 64,
    "learning_starts": 100,
    "sre_num_repeats": 0,
    "sred_max_iters": 50,
    "sred_lr": 0.05,
    "sred_optimizer": "adam",
    "sred_br_temperature": 0.05,
    "sred_gap_temperature": 0.01,
    "sred_gradient_clip_norm": 10.0,
    "sred_eval_every": 10,
    "sred_device": SRED_DEVICE,
}

print(f"SRED hyperparameter overrides: {SRED_HYPERPARAMETER_OVERRIDES}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
NUM_ENVS = 8
EVAL_NUM_ENVS = NUM_ENVS

```

## Robust epsilon 0.01

```python
deep_srq_sred_eps_0_01_stats = train_deepsrq_sred_gradient_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    scenarios=SCENARIOS,
    base_seed=BASE_SEED + 1,
    use_gpu=USE_GPU,
    eval_interval=EVAL_INTERVAL,
    eval_episodes=EVAL_EPISODES_DURING_TRAINING,
    hyperparameter_overrides=SRED_HYPERPARAMETER_OVERRIDES,
    repo_root=REPO_ROOT,
    num_envs=NUM_ENVS,
    eval_num_envs=EVAL_NUM_ENVS,
)
deep_srq_sred_eps_0_01_training_figs = display_training_reward_plots(deep_srq_sred_eps_0_01_stats.values())
deep_srq_sred_eps_0_01_stats
```
