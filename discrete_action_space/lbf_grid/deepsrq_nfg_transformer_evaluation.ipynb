# Deep SRQ NfgTransformer Evaluation


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "lbf_grid":
    REPO_ROOT = ROOT.parents[1]
elif (ROOT / "discrete_action_space" / "lbf_grid").exists():
    REPO_ROOT = ROOT
else:
    REPO_ROOT = ROOT.parents[1]

for path in [REPO_ROOT, REPO_ROOT / "discrete_action_space", REPO_ROOT / "discrete_action_space" / "lbf_grid"]:
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.lbf_grid.robust_notebook_utils import (
    ROBUST_EPSILONS,
    display_evaluation_boxplots,
    evaluate_deepsrq_nfg_transformer_suite_for_epsilon,
    robust_lbf_scenarios,
)

SCENARIOS = robust_lbf_scenarios()
for scenario in SCENARIOS:
    print(f"{scenario.key}: {scenario.name}")
```

```python
EVAL_EPISODES = 50
USE_GPU = True
BASE_SEED = 2025
EVAL_NUM_ENVS = 16
NFG_TRANSFORMER_CHECKPOINT_PATH = (
    REPO_ROOT
    / "discrete_action_space"
    / "sre_solvers"
    / "nfg_transformer"
    / "nfg_sre_checkpoints"
    / "nfg_sre_lbf3_online.pt"
)
NFG_TRANSFORMER_HYPERPARAMETER_OVERRIDES = {
    "agent": {
        "train_every": 4,
        "target_equilibrium_update_steps": 4,
        "sre_policy_cache_enabled": False,
    },
    "nfg_transformer": {
        "checkpoint_path": NFG_TRANSFORMER_CHECKPOINT_PATH,
        "fallback_enabled": False,
        "accept_exploitability_tol": None,
        "compute_exploitability_diagnostics": True,
        "device": "cuda" if USE_GPU else "cpu",
    },
}

print(f"NfgTransformer checkpoint: {NFG_TRANSFORMER_CHECKPOINT_PATH}")
print(f"Checkpoint exists: {NFG_TRANSFORMER_CHECKPOINT_PATH.exists()}")

print(f"Evaluation episodes per matchup: {EVAL_EPISODES}")
print(f"Vectorized evaluation envs: {EVAL_NUM_ENVS}")
print(f"NfgTransformer hyperparameter overrides: {NFG_TRANSFORMER_HYPERPARAMETER_OVERRIDES}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01


```python
deep_srq_nfg_transformer_eval_eps_0_01 = evaluate_deepsrq_nfg_transformer_suite_for_epsilon(
    0.01,
    scenarios=SCENARIOS,
    n_episodes=EVAL_EPISODES,
    repo_root=REPO_ROOT,
    use_gpu=USE_GPU,
    num_envs=EVAL_NUM_ENVS,
    hyperparameter_overrides=NFG_TRANSFORMER_HYPERPARAMETER_OVERRIDES,
)
display_evaluation_boxplots(deep_srq_nfg_transformer_eval_eps_0_01)
```
