# SR-A2C LBF Training

Training-only notebook for SR-A2C with the PATH MCP N-player process pool. Each epsilon cell runs all configured LBF scenarios and saves full-resume checkpoints plus reward plots.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "lbf_grid":
    REPO_ROOT = ROOT.parents[1]
elif (ROOT / "discrete_action_space" / "lbf_grid").exists():
    REPO_ROOT = ROOT
else:
    REPO_ROOT = ROOT.parents[1]

for path in [REPO_ROOT, REPO_ROOT / "discrete_action_space", REPO_ROOT / "discrete_action_space" / "lbf_grid"]:
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.lbf_grid.robust_notebook_utils import (
    DEFAULT_EVAL_EPISODES,
    ROBUST_EPSILONS,
    robust_lbf_scenarios,
    train_sra2c_for_epsilon,
)
from discrete_action_space.lbf_grid.notebook_eval import display_training_reward_plots

SCENARIOS = robust_lbf_scenarios()
for scenario in SCENARIOS:
    print(f"{scenario.key}: {scenario.name}")
```

```python
N_EPISODES = 500
USE_GPU = True
BASE_SEED = 2025
NUM_ENVS = 32
EVAL_NUM_ENVS = 32
EVAL_INTERVAL = 100
EVAL_EPISODES_DURING_TRAINING = 10
SRE_SOLVER_WORKERS = 16
SRA2C_HYPERPARAMETER_OVERRIDES = {
    "agent": {
        "batch_size": 64,
        "train_every": 8,
        "rollout_steps": 8,
        "actor_update_every": 1,
        "sre_num_repeats": 2,
        "sre_include_pure_starts": False,
        "sre_target_value_mode": "nominal",
        "entropy_coef": 0.01,
        "value_loss_coef": 0.5,
        "sre_imitation_coef": 0.05,
        "advantage_normalization": True,
    },
}
print(f"Vectorized training envs: {NUM_ENVS}")
print(f"SR-A2C rollout batch size: {SRA2C_HYPERPARAMETER_OVERRIDES['agent']['batch_size']}")
print(f"PATH MCP N-player pool workers: {SRE_SOLVER_WORKERS}")
print(f"SR-A2C hyperparameter overrides: {SRA2C_HYPERPARAMETER_OVERRIDES}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01

```python
sra2c_eps_0_01_stats = train_sra2c_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    scenarios=SCENARIOS,
    base_seed=BASE_SEED + 1,
    use_gpu=USE_GPU,
    eval_interval=EVAL_INTERVAL,
    eval_episodes=EVAL_EPISODES_DURING_TRAINING,
    sre_solver_workers=SRE_SOLVER_WORKERS,
    num_envs=NUM_ENVS,
    eval_num_envs=EVAL_NUM_ENVS,
    hyperparameter_overrides=SRA2C_HYPERPARAMETER_OVERRIDES,
    repo_root=REPO_ROOT,
)
sra2c_eps_0_01_training_figs = display_training_reward_plots(sra2c_eps_0_01_stats.values())
sra2c_eps_0_01_stats
```
