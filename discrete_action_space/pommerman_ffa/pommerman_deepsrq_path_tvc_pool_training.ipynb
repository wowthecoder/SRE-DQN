# Pommerman FFA Deep SRQ PATH TVC Pool Training

Train full four-agent Deep SRQ runs with the `path_tvc_mcp_nplayer_pool` stage solver.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
for parent in (ROOT, *ROOT.parents):
    if (parent / "discrete_action_space").exists() and (parent / "relevant_papers").exists():
        ROOT = parent
        break

for path in (ROOT, ROOT / "discrete_action_space", ROOT / "discrete_action_space" / "bimatrix_game"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from discrete_action_space.pommerman_ffa.notebook_utils import (
    configure_notebook_imports,
    train_pommerman_deepsrq_path_tvc_pool_for_epsilon,
)

ROOT = configure_notebook_imports(ROOT)
N_EPISODES = 100
MAX_STEPS = 200
BATCH_SIZE = 32
USE_GPU = True
POOL_SIZE = 16
SRE_NUM_REPEATS = 4
SRE_INCLUDE_PURE_STARTS = False
TARGET_EQUILIBRIUM_UPDATE_STEPS = 4
```

```python
stats_eps_001 = train_pommerman_deepsrq_path_tvc_pool_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    max_steps=MAX_STEPS,
    repo_root=ROOT,
    use_gpu=USE_GPU,
    batch_size=BATCH_SIZE,
    sre_solver_workers=POOL_SIZE,
    sre_num_repeats=SRE_NUM_REPEATS,
    sre_include_pure_starts=SRE_INCLUDE_PURE_STARTS,
    target_equilibrium_update_steps=TARGET_EQUILIBRIUM_UPDATE_STEPS,
)
```
