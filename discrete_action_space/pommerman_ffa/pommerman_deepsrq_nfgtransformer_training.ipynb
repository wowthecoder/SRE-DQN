# Pommerman FFA Deep SRQ + NfgTransformer Training

Training-only notebook for full four-agent Pommerman FFA. Runs Deep SRQ with the `nfg_transformer_sre` stage-game solver for each robust epsilon. PATH fallback is disabled by default because the Pommerman FFA stage game is too large for PATH-in-the-loop experiments.

```python
from pathlib import Path
import sys


def _find_repo_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for path in (current, *current.parents):
        if (path / "discrete_action_space").exists() and (path / "relevant_papers").exists():
            return path
    raise RuntimeError(f"Could not find repo root from {current}")

ROOT = _find_repo_root()
for path in (ROOT, ROOT / "discrete_action_space", ROOT / "discrete_action_space" / "bimatrix_game"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

POMMERMAN_DIR = ROOT / "discrete_action_space" / "pommerman_ffa"

from discrete_action_space.pommerman_ffa.notebook_utils import (
    DEFAULT_NFG_TRANSFORMER_CHECKPOINT,
    ROBUST_EPSILONS,
    display_training_reward_plots,
    train_pommerman_deepsrq_nfgtransformer_for_epsilon,
)

N_EPISODES = 100
MAX_STEPS = 200
BATCH_SIZE = 32
BASE_SEED = 2025
USE_GPU = True
NFG_CHECKPOINT_PATH = DEFAULT_NFG_TRANSFORMER_CHECKPOINT
NFG_ACCEPT_GAP = None
NFG_FALLBACK_ENABLED = False

if not Path(NFG_CHECKPOINT_PATH).exists():
    raise FileNotFoundError(f"NfgTransformer checkpoint not found: {NFG_CHECKPOINT_PATH}")

print(f"Pommerman dir: {POMMERMAN_DIR}")
print(f"NfgTransformer checkpoint: {NFG_CHECKPOINT_PATH}")
print(f"NfgTransformer PATH fallback enabled: {NFG_FALLBACK_ENABLED}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01

```python
deep_srq_eps_0_01_stats = train_pommerman_deepsrq_nfgtransformer_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    max_steps=MAX_STEPS,
    seed=BASE_SEED + int(0.01 * 10000),
    repo_root=ROOT,
    use_gpu=USE_GPU,
    nfg_checkpoint_path=NFG_CHECKPOINT_PATH,
    nfg_accept_gap=NFG_ACCEPT_GAP,
    nfg_fallback_enabled=NFG_FALLBACK_ENABLED,
    batch_size=BATCH_SIZE,
)
display_training_reward_plots(deep_srq_eps_0_01_stats)
deep_srq_eps_0_01_stats
```
