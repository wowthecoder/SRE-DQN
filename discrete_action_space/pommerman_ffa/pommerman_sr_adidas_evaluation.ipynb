# Pommerman FFA SR-ADIDAS Evaluation

Loads the best SR-ADIDAS checkpoint for each robust epsilon, runs evaluation rollouts, saves reward boxplots, and saves a sampled rollout GIF when rendering is available.

```python
from pathlib import Path
import sys


def _find_repo_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for path in (current, *current.parents):
        if (path / "discrete_action_space").exists() and (path / "relevant_papers").exists():
            return path
    raise RuntimeError(f"Could not find repo root from {current}")

ROOT = _find_repo_root()
for path in (ROOT, ROOT / "discrete_action_space", ROOT / "discrete_action_space" / "bimatrix_game"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

POMMERMAN_DIR = ROOT / "discrete_action_space" / "pommerman_ffa"

from discrete_action_space.pommerman_ffa.notebook_utils import (
    ROBUST_EPSILONS,
    display_evaluation_rollout,
    evaluate_pommerman_sr_adidas_for_epsilon,
)

EVAL_EPISODES = 20
MAX_STEPS = 200
EVAL_VIDEO_FPS = 4
BASE_SEED = 2025
USE_GPU = True

print(f"Evaluation episodes per epsilon: {EVAL_EPISODES}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01

```python
sr_adidas_eval_eps_0_01 = evaluate_pommerman_sr_adidas_for_epsilon(
    0.01,
    n_episodes=EVAL_EPISODES,
    max_steps=MAX_STEPS,
    seed=BASE_SEED + 60_000 + int(0.01 * 10000),
    repo_root=ROOT,
    use_gpu=USE_GPU,
)
display_evaluation_rollout(sr_adidas_eval_eps_0_01, fps=EVAL_VIDEO_FPS)
sr_adidas_eval_eps_0_01
```
