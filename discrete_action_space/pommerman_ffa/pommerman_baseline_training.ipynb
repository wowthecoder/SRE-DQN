```python
from pathlib import Path
from IPython.display import display

from discrete_action_space.pommerman_ffa.notebook_utils import (
    BASE_SEED,
    baseline_training_dir,
    configure_notebook_imports,
    display_training_reward_plots,
    train_iql_dqn,
    train_ippo,
)

REPO_ROOT = configure_notebook_imports()
OUTPUT_ROOT = baseline_training_dir(repo_root=REPO_ROOT)
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
REPO_ROOT, OUTPUT_ROOT
```

```python
SMOKE_TEST = True
TRAIN_IQL_DQN = True
TRAIN_IPPO = True

N_EPISODES = 8 if SMOKE_TEST else 500
MAX_STEPS = 80 if SMOKE_TEST else 200
BATCH_SIZE = 64
N_TRAIN_ENVS = 2 if SMOKE_TEST else 4
USE_GPU = True
SEED = BASE_SEED

config = {
    "n_episodes": N_EPISODES,
    "max_steps": MAX_STEPS,
    "batch_size": BATCH_SIZE,
    "n_train_envs": N_TRAIN_ENVS,
    "output_root": str(OUTPUT_ROOT),
}
config
```

```python
baseline_stats = {}

if TRAIN_IQL_DQN:
    baseline_stats["iql_dqn"] = train_iql_dqn(
        n_episodes=N_EPISODES,
        max_steps=MAX_STEPS,
        seed=SEED,
        output_root=OUTPUT_ROOT,
        use_gpu=USE_GPU,
        batch_size=BATCH_SIZE,
        n_envs=N_TRAIN_ENVS,
        verbose=True,
    )

if TRAIN_IPPO:
    baseline_stats["ippo"] = train_ippo(
        n_episodes=N_EPISODES,
        max_steps=MAX_STEPS,
        seed=SEED + 1_000,
        output_root=OUTPUT_ROOT,
        use_gpu=USE_GPU,
        batch_size=BATCH_SIZE,
        n_envs=N_TRAIN_ENVS,
        verbose=True,
    )

list(baseline_stats)
```

```python
for name, stats in baseline_stats.items():
    print(name, stats.get("artifact_dir"))
    display_training_reward_plots(stats)
```
