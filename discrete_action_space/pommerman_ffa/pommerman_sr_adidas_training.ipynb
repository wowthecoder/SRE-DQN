# Pommerman FFA SR-ADIDAS Training

Training-only notebook for full four-agent Pommerman FFA. Runs SR-ADIDAS for each robust epsilon with a constant robust-epsilon schedule.

```python
from pathlib import Path
import sys


def _find_repo_root(start=None):
    current = Path(start or Path.cwd()).resolve()
    for path in (current, *current.parents):
        if (path / "discrete_action_space").exists() and (path / "relevant_papers").exists():
            return path
    raise RuntimeError(f"Could not find repo root from {current}")

ROOT = _find_repo_root()
for path in (ROOT, ROOT / "discrete_action_space", ROOT / "discrete_action_space" / "bimatrix_game"):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

POMMERMAN_DIR = ROOT / "discrete_action_space" / "pommerman_ffa"

from discrete_action_space.pommerman_ffa.notebook_utils import (
    ROBUST_EPSILONS,
    display_training_reward_plots,
    train_pommerman_sr_adidas_for_epsilon,
)

N_EPISODES = 100
MAX_STEPS = 200
BATCH_SIZE = 32
BASE_SEED = 2025
USE_GPU = True

print(f"Pommerman dir: {POMMERMAN_DIR}")
print(f"Robust epsilons: {ROBUST_EPSILONS}")
```

## Robust epsilon 0.01

```python
sr_adidas_eps_0_01_stats = train_pommerman_sr_adidas_for_epsilon(
    0.01,
    n_episodes=N_EPISODES,
    max_steps=MAX_STEPS,
    seed=BASE_SEED + 20_000 + int(0.01 * 10000),
    repo_root=ROOT,
    use_gpu=USE_GPU,
    batch_size=BATCH_SIZE,
)
display_training_reward_plots(sr_adidas_eps_0_01_stats)
sr_adidas_eps_0_01_stats
```
